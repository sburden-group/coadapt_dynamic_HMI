# Co-adaptation trials

```python
# 2nd order interface coadaptation
G_parameters2,Gs2,Hs2,Haccs2,Ds2,UHs2,UGs2,Ys2,ds2,uhs2,ugs2,ys2,errors2 = analyze(DATA_PATH,keys,'_2nd_coadapt')

# 1st order interface coadaptation
G_parameters1,Gs1,Hs1,Haccs1,Ds1,UHs1,UGs1,Ys1,ds1,uhs1,ugs1,ys1,errors1 = analyze(DATA_PATH,keys,'_1st_coadapt')

# 0th order interface coadaptation
G_parameters0,Gs0,Hs0,Haccs0,Ds0,UHs0,UGs0,Ys0,ds0,uhs0,ugs0,ys0,errors0 = analyze(DATA_PATH,keys,'_0th_coadapt')
```

```python
# combine the data from the 3 conditions (0th, 1st, 2nd order coadaptation)
G_parameters = [G_parameters0, G_parameters1, G_parameters2] # (different # of parameters)
Gs = np.stack((Gs0, Gs1, Gs2), axis=1)
Hs = np.stack((Hs0, Hs1, Hs2), axis=1)
Ds = np.stack((Ds0, Ds1, Ds2), axis=1)
UHs = np.stack((UHs0, UHs1, UHs2), axis=1)
UGs = np.stack((UGs0, UGs1, UGs2), axis=1)
Ys = np.stack((Ys0, Ys1, Ys2), axis=1)
ds = np.stack((ds0, ds1, ds2), axis=1)
uhs = np.stack((uhs0, uhs1, uhs2), axis=1)
ugs = np.stack((ugs0, ugs1, ugs2), axis=1)
ys = np.stack((ys0, ys1, ys2), axis=1)
errors = np.stack((errors0, errors1, errors2), axis=1)
```

```python
subject_num = len(keys)
condition_num = 3
trial_num = len(Gs2[0])
print('subject_num = ',subject_num)
print('condition_num (orders) = ',condition_num)
print('trial_num = ',trial_num)
print('\n')

# interface transfer function parameters (subjects x trials x number of parameters)
print(G_parameters[0].shape,G_parameters[1].shape,G_parameters[2].shape) 
print('\n')

# freqeuncy domain
print(Gs.shape) # interface transfer function (subjects x condition x trials x stim frequencies)
print(Hs.shape) # human transfer function (subjects x condition x trials x stim frequencies)
print(Ds.shape) # disturbances (subjects x condition x trials x number of frequencies)
print(UHs.shape) # human inputs (subjects x condition x trials x number of frequencies)
print(UGs.shape) # interface inputs (subjects x condition x trials x number of frequencies)
print(Ys.shape) # output (subjects x condition x trials x number of frequencies)
print('\n')

# time domain
print(ds.shape) # disturbances (subjects x condition x trials x number of time points)
print(uhs.shape) # human inputs (subjects x condition x trials x number of time points)
print(ugs.shape) # interface inputs (subjects x condition x trials x number of time points)
print(ys.shape) # output (subjects x condition x trials x number of time points)
print(errors.shape) # errors (subjects x condition x trials)
```
